```python


# Number of trials
n = 10000

# Simulate coin toss (Binomial with 1 trial)
coin_tosses = np.random.binomial(n=1, p=0.5, size=n)

# Compute running mean
running_mean = np.cumsum(coin_tosses) / np.arange(1, n+1)

print("Final Average:", running_mean[-1])
```

```python
plt.plot(running_mean)
plt.axhline(y=0.5)
plt.title("Law of Large Numbers - Coin Toss")
plt.xlabel("Number of Trials")
plt.ylabel("Running Average")
plt.show()
```

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 4, 5])

cov_matrix = np.cov(x, y)
print("Covariance Matrix:\n", cov_matrix)
```

```python
df = pd.DataFrame({
    "Hours": [1, 2, 3, 4, 5],
    "Scores": [2, 4, 5, 4, 5]
})

print(df.cov())
```

```python
corr_matrix = np.corrcoef(x, y)
print("Correlation Matrix:\n", corr_matrix)
```
